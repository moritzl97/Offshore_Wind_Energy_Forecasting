```python
import pandas as pd
import json

# Load metadata
with open("../metadata_wind_farms.json") as f:
    metadata = json.load(f)

# Check each farm
for farm in metadata["wind_farms"]:
    if not farm.get("offshore_data_available", False):
        continue
    
    farm_id      = farm["id"]
    knmi_station = farm["nearest_KNMI_station"]
    
    # load offshore
    offshore = pd.read_csv(f"../datasets/offshore_{farm_id}.csv", 
                           parse_dates=["time"])
    
    # load knmi
    knmi = pd.read_csv(f"../datasets/knmi_{knmi_station}.csv", 
                       parse_dates=["timestamp"])
    
    print(f"Farm: {farm_id}")
    print(f"  Offshore: {offshore['time'].min()} → {offshore['time'].max()}")
    print(f"  KNMI:     {knmi['timestamp'].min()} → {knmi['timestamp'].max()}")
    print()
```
