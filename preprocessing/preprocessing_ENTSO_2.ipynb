# ENTSO Reference and Yearly Alignment

This notebook stores the ENTSO offshore installed capacity reference and aligns it with the Kaggle yearly wind power summary.

```python
import pandas as pd
from pathlib import Path

DATA_DIR = Path('../datasets')
RAW_ENTSO_DIR = Path('../datasets_raw/ENTSO_reference')
```

```python
entso = pd.DataFrame({
    'year': list(range(2015, 2027)),
    'total_installed_capacity_MW': [228.0, 957.0, 957.0, 957.0, 957.0, 2460.0, 2460.0, 2460.0, 4739.0, 4739.0, 4739.0, 4739.0],
    'Borssele_34': [0,0,0,0,0,0,0,731.5,731.5,731.5,731.5,731.5],
    'Hollandse_Kust_Zuid': [0,0,0,0,0,0,0,0,1420.0,1420.0,1420.0,1420.0],
    'Hollandse_Kust_Noord': [0,0,0,0,0,0,0,0,0,760.0,760.0,760.0],
    'Gemini': [0,0,0,600,600,600,600,600,600,600,600,600],
    'Borssele_12': [0,0,0,0,0,0,752,752,752,752,752,752],
    'Others': [108,252,252,252,252,252,252,501,501,501,501,501]
})

entso.to_csv(DATA_DIR / 'ENTSO_installed_capacity_offshore.csv', index=False)
entso.to_csv(RAW_ENTSO_DIR / 'ENTSO_installed_capacity_offshore_reference.csv', index=False)
display(entso)
```

```python
wind = pd.read_csv(DATA_DIR / 'wind_turbine_train_clean.csv')
annual_power = wind.groupby('year', as_index=False).agg(
    kaggle_avg_power=('Power','mean'),
    kaggle_avg_wind_speed_100m=('WS_100m','mean'),
    kaggle_avg_temperature_2m=('Temp_2m','mean'),
    kaggle_rows=('Power','size')
)

display(annual_power)
```

```python
merged = pd.merge(annual_power, entso, on='year', how='left')
merged.to_csv(DATA_DIR / 'merged_kaggle_entso_yearly.csv', index=False)
display(merged)
```
