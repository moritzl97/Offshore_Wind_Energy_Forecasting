```python
import os
import pandas as pd

BASE_DIR    = os.path.dirname(os.getcwd())
FOLDER_PATH = os.path.join(BASE_DIR, "datasets")

def search_csv(search_term, find_max=False, find_min=False, find_median=False, filename_filter=None):
    results = []

    for filename in os.listdir(FOLDER_PATH):
        if not filename.endswith(".csv"):
            continue

        if filename_filter and filename_filter.lower() not in filename.lower():
            continue

        df = pd.read_csv(os.path.join(FOLDER_PATH, filename))

        for col in df.columns:
            if search_term.lower() in col.lower():
                numeric_col = pd.to_numeric(df[col], errors='coerce')

                if find_max or find_min or find_median:
                    row = {"file": filename, "column": col}
                    if find_max:
                        row["max"]    = numeric_col.max()
                    if find_min:
                        row["min"]    = numeric_col.min()
                    if find_median:
                        row["median"] = numeric_col.median()
                    results.append(row)
                else:
                    for _, r in df.iterrows():
                        results.append({
                            "file"  : filename,
                            "column": col,
                            "value" : r[col]
                        })

    if not results:
        print(f"No matches found for '{search_term}'")
        return
    
    result_df = pd.DataFrame(results)

    if not find_max and not find_min and not find_median:
        total = len(result_df)
        result_df = result_df.head(20)
        print(f"Showing {len(result_df)} of {total} results:\n")
        
    print(result_df.to_string(index=False))
```

```python
search_csv("Windspeed")
search_csv("Windspeed", find_max=True)
search_csv("Windspeed", find_min=True)
search_csv("Windspeed", find_median=True)

search_csv("Windspeed", find_max=True, find_min=True, find_median=True)

search_csv("Windspeed", find_max=True, filename_filter="offshore_Gemini")

search_csv("windspeed", find_max=True, find_min=True, find_median=True, filename_filter="Borssele_12")
```
